```python
# Select the backend before importing Keras. Restart the kernel after changing it.
import os

os.environ["KERAS_BACKEND"] = "torch"

import keras
import torch

import numpy as np
```

```python
y_true = np.array([0, 2], dtype="int64")
y_pred = np.array(
        [[0.8, 0.1, 0.1],
         [0.2, 0.3, 0.5]], dtype="float32"
        )
y_one_hot = keras.utils.to_categorical(y_true, num_classes=3)
print(y_one_hot)
```

```python
sparse_loss = keras.losses.SparseCategoricalCrossentropy()
one_hot_loss = keras.losses.CategoricalCrossentropy()
print(keras.ops.convert_to_numpy(sparse_loss(y_true, y_pred)))
print(keras.ops.convert_to_numpy(one_hot_loss(y_one_hot, y_pred)))
```

```python
# Define Sequential model with 3 layers
from keras import layers

model = keras.Sequential(
        [
                layers.Dense(2, activation="relu", name="layer1"),
                layers.Dense(3, activation="relu", name="layer2"),
                layers.Dense(4, activation="softmax", name="layer3"),
                ]
        )
# Call model on a test input
x = torch.rand((5, 3))
print(x)
y = model(x)
print(y)
```

```python
model = keras.Sequential(
        [
                keras.Input(shape=(28, 28)),
                keras.layers.Flatten(),
                keras.layers.Dense(20, activation="relu"),
                keras.layers.Dense(10, activation="softmax")
                ]
        )
model.summary()
```

```python
img = torch.rand((4, 28, 28))
y = model.predict(img)
print(y.shape)
```
